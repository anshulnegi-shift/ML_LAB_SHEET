# 33. Create a histogram for a numerical feature using Matplotlib. 

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

iris = load_iris()

plt.hist(iris.data[:, 0], bins=10)
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.title("Histogram of Sepal Length")
plt.show()
```

# 34. Create a scatter plot to visualize the relationship between two variables. 

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

iris = load_iris()

plt.scatter(
    iris.data[:, 0],
    iris.data[:, 1]
)

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Scatter Plot")
plt.show()
```

# 35. Generate a correlation matrix and visualize it using a heatmap

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

correlation = df.corr()

print(correlation)

sns.heatmap(
    correlation,
    annot=True
)

plt.title("Correlation Heatmap")
plt.show()
```
