```python
import pandas as pd

df = pd.read_csv("student_data.csv")

print(df)
```

# 1. Load a dataset and identify missing values in each column. 

```python
print(df.isnull().sum())
```

# 2. Display the percentage of missing values in every feature. 

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

# 3. Remove rows containing missing values from the dataset. 

```python
df_no_missing_rows = df.dropna()

print(df_no_missing_rows)
```

# 4. Remove columns having more than 50% missing values. 

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

columns_to_remove = missing_percentage[missing_percentage > 50].index

df_removed_columns = df.drop(columns=columns_to_remove)

print(df_removed_columns)
```

# 5. Replace missing numerical values using the mean. 

```python
df_mean = df.copy()

numeric_columns = df_mean.select_dtypes(include="number").columns

for column in numeric_columns:
    df_mean[column] = df_mean[column].fillna(df_mean[column].mean())

print(df_mean)
```

# 6. Replace missing numerical values using the median.

```python
df_median = df.copy()

numeric_columns = df_median.select_dtypes(include="number").columns

for column in numeric_columns:
    df_median[column] = df_median[column].fillna(df_median[column].median())

print(df_median)
```

# 7. Replace missing categorical values using the mode. 

```python
df_mode = df.copy()

categorical_columns = df_mode.select_dtypes(include="str").columns

for column in categorical_columns:
    df_mode[column] = df_mode[column].fillna(df_mode[column].mode()[0])

print(df_mode)
```

# 8. Fill missing values using forward fill (Forward Propagation). 

```python
df_ffill = df.copy()

df_ffill = df_ffill.ffill()

print(df_ffill)
```

# 9. Fill missing values using backward fill (Backward Propagation). 

```python
df_bfill = df.copy()

df_bfill = df_bfill.bfill()

print(df_bfill)
```

# 10. Compare the dataset before and after handling missing values. 

```python
before = df.isnull().sum().sum()

after = df_mean.isnull().sum().sum()

print("Missing values before handling:", before)
print("Missing values after handling:", after)
```

# 11.  Detect outliers using the Interquartile Range (IQR) method. 

```python
Q1 = df["Marks"].quantile(0.25)
Q3 = df["Marks"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Marks"] < lower_limit) |
    (df["Marks"] > upper_limit)
]

print("Outliers:")
print(outliers)
```

# 12. Detect outliers using the Z-score method. 

```python
mean = df["Marks"].mean()
std = df["Marks"].std()

df["Z_Score"] = (df["Marks"] - mean) / std

outliers = df[abs(df["Z_Score"]) > 3]

print("Outliers:")
print(outliers)
```

# 14. Visualize outliers using a Scatter Plot. 

```python
plt.scatter(df.index, df["Marks"])

plt.title("Scatter Plot of Marks")
plt.xlabel("Index")
plt.ylabel("Marks")

plt.show()
```

# 15. Remove outliers using the IQR method. 

```python
Q1 = df["Marks"].quantile(0.25)
Q3 = df["Marks"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df_iqr = df[
    (df["Marks"] >= lower_limit) &
    (df["Marks"] <= upper_limit)
]

print(df_iqr)
```

# 16. Replace outliers with the median value. 

```python
df_median_outlier = df.copy()

Q1 = df_median_outlier["Marks"].quantile(0.25)
Q3 = df_median_outlier["Marks"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

median_value = df_median_outlier["Marks"].median()

outlier_condition = (
    (df_median_outlier["Marks"] < lower_limit) |
    (df_median_outlier["Marks"] > upper_limit)
)

df_median_outlier.loc[outlier_condition, "Marks"] = median_value

print(df_median_outlier)
```

# 17. Cap outliers using percentile-based capping. 

```python
df_capped = df.copy()

lower = df_capped["Marks"].quantile(0.05)
upper = df_capped["Marks"].quantile(0.95)

df_capped["Marks"] = df_capped["Marks"].clip(
    lower=lower,
    upper=upper
)

print(df_capped)
```

# 18. Compare the dataset before and after outlier treatment. 

```python
print("Before outlier treatment:")
print(df["Marks"].describe())

print("\nAfter outlier treatment:")
print(df_iqr["Marks"].describe())
```

# 19. Apply Min-Max Normalization to numerical features. 

```python
from sklearn.preprocessing import MinMaxScaler

df_minmax = df.copy()

scaler = MinMaxScaler()

df_minmax[["Age", "Marks"]] = scaler.fit_transform(
    df_minmax[["Age", "Marks"]].fillna(
        df_minmax[["Age", "Marks"]].mean()
    )
)

print(df_minmax)
```

# 20. Apply Standardization (Z-score Scaling). 

```python
from sklearn.preprocessing import StandardScaler

df_standard = df.copy()

scaler = StandardScaler()

df_standard[["Age", "Marks"]] = scaler.fit_transform(
    df_standard[["Age", "Marks"]].fillna(
        df_standard[["Age", "Marks"]].mean()
    )
)

print(df_standard)
```

# 21. Apply Robust Scaling to handle outliers. 

```python
from sklearn.preprocessing import RobustScaler

df_robust = df.copy()

scaler = RobustScaler()

df_robust[["Age", "Marks"]] = scaler.fit_transform(
    df_robust[["Age", "Marks"]].fillna(
        df_robust[["Age", "Marks"]].median()
    )
)

print(df_robust)
```

# 22. Apply Max Absolute Scaling. 

```python
from sklearn.preprocessing import MaxAbsScaler

df_maxabs = df.copy()

scaler = MaxAbsScaler()

df_maxabs[["Age", "Marks"]] = scaler.fit_transform(
    df_maxabs[["Age", "Marks"]].fillna(
        df_maxabs[["Age", "Marks"]].mean()
    )
)

print(df_maxabs)
```

# 27. Encode categorical variables using Label Encoding. 

```python
from sklearn.preprocessing import LabelEncoder

df_label = df.copy()

df_label["City"] = df_label["City"].fillna(
    df_label["City"].mode()[0]
)

encoder = LabelEncoder()

df_label["City_Encoded"] = encoder.fit_transform(
    df_label["City"]
)

print(df_label)
```

# 28. Encode categorical variables using One-Hot Encoding. 

```python
df_onehot = df.copy()

df_onehot["City"] = df_onehot["City"].fillna(
    df_onehot["City"].mode()[0]
)

df_onehot = pd.get_dummies(
    df_onehot,
    columns=["City"],
    dtype=int
)

print(df_onehot)
```

# 30. Create a new feature by combining two existing columns. 

```python
df_feature = df.copy()

df_feature["Age_Marks"] = (
    df_feature["Age"].fillna(df_feature["Age"].mean()) +
    df_feature["Marks"].fillna(df_feature["Marks"].mean())
)

print(df_feature)
```

# 31. Extract year, month, and day from a date column.

# 33. Apply Log Transformation to skewed data. 

```python
import numpy as np
```

# 35. Create a final preprocessed dataset ready for Machine Learning.

```python
final_df = df.copy()

numeric_columns = final_df.select_dtypes(
    include="number"
).columns

for column in numeric_columns:
    final_df[column] = final_df[column].fillna(
        final_df[column].mean()
    )

categorical_columns = final_df.select_dtypes(
    include="str"
).columns

for column in categorical_columns:
    final_df[column] = final_df[column].fillna(
        final_df[column].mode()[0]
    )

final_df = pd.get_dummies(
    final_df,
    columns=["City"],
    dtype=int
)

print(final_df)
```
