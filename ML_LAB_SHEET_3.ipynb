# 13. Implement Multiple Linear Regression using multiple independent variables. 

```python
X_multi = df[[
    "MedInc",
    "HouseAge",
    "AveRooms",
    "Population"
]]

y_multi = df["Price"]

print(X_multi.head())
```

# 14. Train the Multiple Linear Regression model. 

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X_train_multi, X_test_multi, y_train_multi, y_test_multi = train_test_split(
    X_multi,
    y_multi,
    test_size=0.2,
    random_state=42
)

multi_model = LinearRegression()
multi_model.fit(X_train_multi, y_train_multi)

print("Multiple Linear Regression model trained successfully")
```

# 15. Predict output values using the testing dataset. 

```python
y_pred_multi = multi_model.predict(X_test_multi)

print("First 10 predictions:")
print(y_pred_multi[:10].round(3))
```

# 16. Compare actual and predicted values graphically. 

```python
import matplotlib.pyplot as plt

plt.scatter(y_test_multi, y_pred_multi, alpha=0.5)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted House Prices")
plt.show()
```

# 17. Analyze the effect of each independent variable on prediction. 

```python
coefficients = pd.DataFrame({
    "Variable": X_multi.columns,
    "Coefficient": multi_model.coef_
})

print(coefficients.round(4))
```

# 18. Implement Polynomial Regression of Degree 2.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

X_poly = df[["MedInc"]]
y_poly = df["Price"]

X_train_poly, X_test_poly, y_train_poly, y_test_poly = train_test_split(
    X_poly,
    y_poly,
    test_size=0.2,
    random_state=42
)

poly2_model = make_pipeline(
    PolynomialFeatures(degree=2),
    LinearRegression()
)

poly2_model.fit(X_train_poly, y_train_poly)

print("Degree 2 Polynomial Regression trained successfully")
```

# 19. Implement Polynomial Regression of Degree 3. 

```python
poly3_model = make_pipeline(
    PolynomialFeatures(degree=3),
    LinearRegression()
)

poly3_model.fit(X_train_poly, y_train_poly)

print("Degree 3 Polynomial Regression trained successfully")
```

# 20. Compare Linear Regression and Polynomial Regression models. 

```python
from sklearn.metrics import mean_squared_error, r2_score

linear_model = LinearRegression()
linear_model.fit(X_train_poly, y_train_poly)

pred_linear = linear_model.predict(X_test_poly)
pred_poly2 = poly2_model.predict(X_test_poly)
pred_poly3 = poly3_model.predict(X_test_poly)

results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Polynomial Degree 2",
        "Polynomial Degree 3"
    ],
    "MSE": [
        mean_squared_error(y_test_poly, pred_linear),
        mean_squared_error(y_test_poly, pred_poly2),
        mean_squared_error(y_test_poly, pred_poly3)
    ],
    "R2 Score": [
        r2_score(y_test_poly, pred_linear),
        r2_score(y_test_poly, pred_poly2),
        r2_score(y_test_poly, pred_poly3)
    ]
})

print(results.round(4))
```

# 21. Visualize Polynomial Regression curves. 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

X_curve = np.linspace(
    X_poly["MedInc"].min(),
    X_poly["MedInc"].max(),
    300
)

X_curve_df = pd.DataFrame({
    "MedInc": X_curve
})
plt.scatter(
    X_train_poly["MedInc"],
    y_train_poly,
    alpha=0.15,
    label="Actual Data"
)
plt.plot(
    X_curve,
    linear_model.predict(X_curve_df),
    label="Linear Regression"
)
plt.plot(
    X_curve,
    poly2_model.predict(X_curve_df),
    label="Polynomial Degree 2"
)
plt.plot(
    X_curve,
    poly3_model.predict(X_curve_df),
    label="Polynomial Degree 3"
)

plt.xlabel("Median Income")
plt.ylabel("House Price")
plt.title("Linear vs Polynomial Regression")
plt.legend()
plt.show()
```

# 23. Compare prediction accuracy for different polynomial degrees. 

```python
degrees = [1, 2, 3, 4, 5]
accuracy_results = []

for degree in degrees:
    poly_model = make_pipeline(
        PolynomialFeatures(degree=degree),
        LinearRegression()
    )

    poly_model.fit(X_train_poly, y_train_poly)

    predictions = poly_model.predict(X_test_poly)

    accuracy_results.append({
        "Degree": degree,
        "MSE": mean_squared_error(y_test_poly, predictions),
        "R2 Score": r2_score(y_test_poly, predictions)
    })

accuracy_df = pd.DataFrame(accuracy_results)

print(accuracy_df.round(4))
```

# 28. Compare the performance of Linear and Polynomial Regression using evaluation metrics. 

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

results = []

models = {
    "Linear Regression": linear_model,
    "Polynomial Degree 2": poly2_model,
    "Polynomial Degree 3": poly3_model
}

for name, model in models.items():

    predictions = model.predict(X_test_poly)

    mae = mean_absolute_error(y_test_poly, predictions)
    mse = mean_squared_error(y_test_poly, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test_poly, predictions)

    results.append([
        name, mae, mse, rmse, r2
    ])

results_df = pd.DataFrame(
    results,
    columns=["Model", "MAE", "MSE", "RMSE", "R2"]
)

print(results_df.round(4))
```

# 31. Train the regression model using standardized features. 

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_multi)
X_test_scaled = scaler.transform(X_test_multi)

scaled_model = LinearRegression()

scaled_model.fit(
    X_train_scaled,
    y_train_multi
)

y_pred_scaled = scaled_model.predict(X_test_scaled)

print("Model trained using standardized features")
```

# 32. Compare model performance before and after feature scaling. 

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np

mae_before = mean_absolute_error(
    y_test_multi,
    y_pred_multi
)

mse_before = mean_squared_error(
    y_test_multi,
    y_pred_multi
)

r2_before = r2_score(
    y_test_multi,
    y_pred_multi
)

mae_after = mean_absolute_error(
    y_test_multi,
    y_pred_scaled
)

mse_after = mean_squared_error(
    y_test_multi,
    y_pred_scaled
)

r2_after = r2_score(
    y_test_multi,
    y_pred_scaled
)

comparison = pd.DataFrame({
    "Metric": ["MAE", "MSE", "R2"],
    "Before Scaling": [
        mae_before,
        mse_before,
        r2_before
    ],
    "After Scaling": [
        mae_after,
        mse_after,
        r2_after
    ]
})

print(comparison.round(4))
```
