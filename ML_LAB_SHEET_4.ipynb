# 1. Load a dataset suitable for clustering using the Pandas library. 

```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)

print(df)
```

# 2. Display the first and last five records of the dataset. 

```python
print("First 5 records:")
print(df.head())

print("\nLast 5 records:")
print(df.tail())
```

# 3. Explore dataset information, summary statistics, and data types. 

```python
print(df.info())

print("\nSummary Statistics:")
print(df.describe())

print("\nData Types:")
print(df.dtypes)
```

# 4. Select relevant numerical features for clustering. 

```python
X = df[[
    "sepal length (cm)",
    "sepal width (cm)",
    "petal length (cm)",
    "petal width (cm)"
]]

print(X.head())
```

# 5. Standardize the dataset before clustering using StandardScaler. 

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print(X_scaled[:5])
```

# 6. Implement the K-Means clustering algorithm. 

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)

kmeans.fit(X_scaled)

print("K-Means model trained successfully.")
```

# 7. Apply K-Means clustering with K = 2 clusters.

```python
df["Cluster"] = kmeans.labels_

print(df.head(10))
```

# 8. Apply K-Means clustering with K = 3 clusters. 

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

kmeans.fit(X)

print("Cluster Labels:")
print(kmeans.labels_)
```

# 9. Apply K-Means clustering with different values of K and compare the results. 

```python
from sklearn.cluster import KMeans

for k in [2, 3, 4, 5]:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)

    print("K =", k)
    print("Inertia =", kmeans.inertia_)
    print()
```

# 10. Determine the optimal number of clusters using the Elbow Method.

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

inertia = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.plot(range(1, 11), inertia, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

# 11. Visualize clusters using a scatter plot.

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

df["Cluster"] = kmeans.fit_predict(X_scaled)

plt.scatter(
    df["petal length (cm)"],
    df["petal width (cm)"],
    c=df["Cluster"]
)

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("K-Means Clusters")
plt.show()
```

# 12. Display the cluster centroids obtained from the K-Means algorithm. 

```python
print("Cluster Centroids:")
print(kmeans.cluster_centers_)
```

# 13. Assign cluster labels to the original dataset. 

```python
df["Cluster"] = kmeans.labels_

print(df.head(10))
```

# 14. Compare clustering results before and after feature scaling.

```python
kmeans_unscaled = KMeans(n_clusters=3, random_state=42, n_init=10)

labels_unscaled = kmeans_unscaled.fit_predict(X)

print("Unscaled Cluster Labels:")
print(labels_unscaled[:10])
```

# 15. Implement Agglomerative Hierarchical Clustering.

```python
from sklearn.cluster import AgglomerativeClustering

hc = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

hc_labels = hc.fit_predict(X_scaled)

print(hc_labels[:10])
```

# 16. Generate a dendrogram using the SciPy library.

```python
from scipy.cluster.hierarchy import dendrogram, linkage

linked = linkage(X_scaled, method="ward")

plt.figure(figsize=(10, 6))

dendrogram(linked)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
```

# 17. Apply Hierarchical Clustering using Ward linkage. 

```python
ward_model = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

ward_labels = ward_model.fit_predict(X_scaled)

print("Ward Labels:")
print(ward_labels[:10])
```

# 18. Apply Hierarchical Clustering using Complete linkage. 

```python
complete_model = AgglomerativeClustering(
    n_clusters=3,
    linkage="complete"
)

complete_labels = complete_model.fit_predict(X_scaled)

print("Complete Linkage Labels:")
print(complete_labels[:10])
```

# 19. Compare different linkage methods. 

```python
methods = ["ward", "complete", "average", "single"]

for method in methods:
    model = AgglomerativeClustering(
        n_clusters=3,
        linkage=method
    )

    labels = model.fit_predict(X_scaled)

    print(method, ":", labels[:10])
```

# 20. Visualize Hierarchical Clustering results. 

```python
plt.scatter(
    df["petal length (cm)"],
    df["petal width (cm)"],
    c=ward_labels
)

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Hierarchical Clustering")
plt.show()
```

# 21. Compare Hierarchical Clustering with K-Means clustering. 

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.scatter(
    X["petal length (cm)"],
    X["petal width (cm)"],
    c=kmeans.labels_
)

plt.title("K-Means")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")


plt.subplot(1, 2, 2)

plt.scatter(
    X["petal length (cm)"],
    X["petal width (cm)"],
    c=ward_labels
)

plt.title("Hierarchical Clustering")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")

plt.show()
```

# 22. Apply Principal Component Analysis (PCA) to reduce dataset dimensions.

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print(X_pca[:5])
```

# 23. Reduce a dataset from multiple features to two principal components. 

```python
pca_df = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

print(pca_df.head())
```

# 24. Visualize the transformed data using a two-dimensional scatter plot. 

```python
plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"]
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA - Two Dimensional Visualization")
plt.show()
```

# 25. Display the explained variance ratio of principal components. 

```python
print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)
```

# 26. Determine the cumulative explained variance. 

```python
import numpy as np

explained_variance = pca.explained_variance_ratio_

cumulative_variance = np.cumsum(explained_variance)

print("Explained Variance Ratio:")
print(explained_variance)

print("\nCumulative Explained Variance:")
print(cumulative_variance)
```

# 27. Compare the dataset before and after dimensionality reduction. 

```python
print("Before PCA:")
print(X.shape)
```

# 28. Perform K-Means clustering on the PCA-transformed dataset.

```python
kmeans_pca = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

pca_labels = kmeans_pca.fit_predict(X_pca)

print("PCA K-Means Labels:")
print(pca_labels[:10])
```

# 29. Compare clustering performance before and after applying PCA. 

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

kmeans_scaled = KMeans(n_clusters=3, random_state=42, n_init=10)

labels_scaled = kmeans_scaled.fit_predict(X_scaled)

score_before = silhouette_score(X_scaled, labels_scaled)

score_after = silhouette_score(X_pca, pca_labels)

print("Silhouette Score Before PCA:", score_before)
print("Silhouette Score After PCA:", score_after)
```

# 30. Calculate the Silhouette Score for cluster evaluation. 

```python
score = silhouette_score(X_scaled, labels_scaled)

print("Silhouette Score:", score)
```

# 31. Compare Silhouette Scores for different values of K. 

```python
for k in range(2, 7):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    cluster_labels = model.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, cluster_labels)

    print("K =", k, "Silhouette Score =", score)
```

# 32. Analyze the characteristics of each cluster using descriptive statistics. 

```python
cluster_analysis = df.groupby("Cluster")[
    [
        "sepal length (cm)",
        "sepal width (cm)",
        "petal length (cm)",
        "petal width (cm)"
    ]
].mean()

print(cluster_analysis)
```
